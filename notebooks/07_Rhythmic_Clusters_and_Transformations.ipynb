# Lab 7 — Rhythmic clusters: sets, cosets, and coherent transformation

A cyclic rhythm can be represented as a subset of $\mathbb Z_N$. Translation produces a coset/rotated set. But musically we often hear **several components as one moving cluster**. This lab makes that idea visual.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def wheel(ax,N,components,title=''):
    th=2*np.pi*np.arange(N)/N
    ax.scatter(np.cos(th),np.sin(th),s=10,alpha=.25)
    for name,S,marker in components:
        a=th[np.array(sorted(S))%N]
        ax.scatter(np.cos(a),np.sin(a),s=80,marker=marker,label=name)
    ax.set_aspect('equal'); ax.axis('off'); ax.set_title(title); ax.legend(loc='upper right',fontsize=8)

def translate(S,g,N): return {(s+g)%N for s in S}
N=28
H7=set(range(0,N,4))       # 7 events
H4=set(range(0,N,7))       # 4 events
C4=translate(H4,5,N)
fig,ax=plt.subplots(figsize=(6,6)); wheel(ax,N,[('7-component',H7,'o'),('4-component shifted by 5',C4,'s')],'A 7:4 cluster on Z_28'); plt.show()
```

## 1. Global rotation versus internal displacement

A global rotation moves the entire cluster while preserving all internal relations. Translating only one component changes the internal phase relation and can change the perceived groove.

```python
fig,axes=plt.subplots(1,3,figsize=(15,5))
wheel(axes[0],N,[('H7',H7,'o'),('5+H4',C4,'s')],'original cluster')
wheel(axes[1],N,[('H7+4',translate(H7,4,N),'o'),('5+H4+4',translate(C4,4,N),'s')],'global rotation +4')
wheel(axes[2],N,[('H7',H7,'o'),('9+H4',translate(H4,9,N),'s')],'internal displacement')
plt.show()
```

## 2. Contraction and expansion are not always group operations

Translation is exact inside a fixed cyclic group. Contraction and expansion generally require changing the embedding/grid or mapping onset positions before requantization. That distinction is useful: it shows where the simple $\mathbb Z_N$ model ends and a richer model of rhythmic transformation begins.

```python
def affine_requantize(S,a,b,N):
    return {int(round(a*s+b))%N for s in S}
fig,axes=plt.subplots(1,3,figsize=(15,5))
wheel(axes[0],N,[('cluster A',H7,'o'),('cluster B',C4,'s')],'reference')
wheel(axes[1],N,[('A contracted',affine_requantize(H7,.8,2,N),'o'),('B contracted',affine_requantize(C4,.8,2,N),'s')],'contracted embedding')
wheel(axes[2],N,[('A expanded',affine_requantize(H7,1.15,-2,N),'o'),('B expanded',affine_requantize(C4,1.15,-2,N),'s')],'expanded embedding')
plt.show()
```
