# CMRM Virtual Rhythm Lab — Start here

This collection accompanies Chapter 3 of **Computer Music Representations and Models**. Each lab follows the same pedagogical pattern:

$$
\boxed{\text{minimal synthetic example}}\to
\boxed{\text{representation}}\to
\boxed{\text{controlled perturbation}}\to
\boxed{\text{interpretation}}\to
\boxed{\text{musical case}}.
$$

The labs are designed to run in Jupyter or Google Colab. No commercial recording is bundled. The *Sound of Muzak* lab uses a synthetic reconstruction of the drum pattern derived from material originally developed by **Bruno Di Giorgi** during his PhD work at Politecnico di Milano. The APM geometry lab incorporates and expands pedagogical material developed by **Riccardo Gianpiccolo**.

## Suggested order

1. Onset detection — *what counts as an onset?*
2. Autocorrelation — *what counts as periodicity?*
3. APM anatomy — *where does phase go, and why does it matter?*
4. Rhythmogram and tempo path — *how does periodicity evolve?*
5. Beat tracking and multiple hypotheses — *period versus phase and ambiguity*
6. *Sound of Muzak* end-to-end — *one musical object through all representations*
7. Rhythmic clusters — *how can several components behave as one object?*

## Provenance

The goal is not to reproduce old research software byte-for-byte. The numerical demonstrations are modern pedagogical reimplementations whose scientific lineage is documented. Bruno Di Giorgi has given permission for his PhD-era material to be reused and modernized for this project. Before public release, the repository should carry a final explicit open-source license and a citation file.
